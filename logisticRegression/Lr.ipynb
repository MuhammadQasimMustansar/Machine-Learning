```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df['species'].unique()
```

```python
df = df[df['species'] != 'virginica']
```

```python
df['species'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder

Label_Encoder = {}

le = LabelEncoder()
df['species'] = le.fit_transform(df['species'])

Label_Encoder["species"] = le

df.head()
```

```python
le.classes_
```

```python
# spliting data into X and y 
# dependent and independent variables

X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, train_size= 0.8, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
```

```python
from sklearn.model_selection import GridSearchCV
parameter={
    'penalty': ['l1','l2','elasticent'],
    'C': [1,2,3,4,5,6,7,8,12,10,20,30],
    'max_iter': [100,200,300]
}
```

```python
classifierRegressor = GridSearchCV(lr,param_grid=parameter,scoring='accuracy',cv=5)
classifierRegressor.fit(X_train,y_train)
```

```python
classifierRegressor.best_params_
```

```python
print(classifierRegressor.best_score_)
```

```python
y_pred = classifierRegressor.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,f1_score,classification_report

print(f'Accuracy Score is: {accuracy_score(y_pred,y_test)}')
print(f'f1 score is: {f1_score(y_pred,y_test)}')
```

```python
print(classification_report(y_test, y_pred))
```
